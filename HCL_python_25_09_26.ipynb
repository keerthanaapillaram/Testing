```python
def longest_unique(arr):
    seen = set()
    left = 0
    max_len = 0
    for right in range(len(arr)):
        while arr[right] in seen:
            seen.remove(arr[left])
            left += 1
        seen.add(arr[right])
        max_len = max(max_len, right - left + 1)
    return max_len
n = int(input("Enter number of students: "))
arr = list(map(int, input("Enter student IDs: ").split()))
print("Longest unique sequence:", longest_unique(arr))
```

```python
n = int(input("\nEnter number of discount values: "))
arr = list(map(int, input("Enter discount values:").split()))
maximum = arr[0]
for i in range(n):
    total = 0
    for j in range(i, n):
        total += arr[j]
        if total > maximum:
            maximum = total
print("Maximum sum:", maximum)
```

```python
n = int(input("\nEnter number of buildings: "))
height = list(map(int, input("Enter building heights:").split()))
water = 0
for i in range(n):
    left = max(height[:i + 1])
    right = max(height[i:])
    water += min(left, right) - height[i]
print("Water trapped:", water)
```

```python
n = int(input("\nEnter number of sales values: "))
sales = list(map(int, input("Enter sales values: ").split()))
maximum = sales[0]
for i in range(n):
    product = 1
    for j in range(i, n):
        product *= sales[j]
        if product > maximum:
            maximum = product
print("Maximum product:", maximum)
```

```python
n = int(input("\nEnter number of purchases: "))
products = list(map(int, input("Enter product IDs:").split()))
longest = 0
for i in range(n):
    unique = []
    for j in range(i, n):
        if products[j] in unique:
            break
        unique.append(products[j])
        if len(unique) > longest:
            longest = len(unique)
print("Longest unique purchase sequence:", longest)
```

```python
n = int(input("\nEnter number of transactions: "))
transactions = list(map(int, input("Enter transaction amounts: ").split()))
target = int(input("Enter target amount: "))
count = 0
for i in range(n):
    total = 0
    for j in range(i, n):
        total += transactions[j]
        if total == target:
            count += 1
print("Number of groups:", count)
```

```python
n = int(input("\nEnter number of employees: "))
skills = input("Enter skill codes: ").split()
groups = {}
for word in skills:
    key = ''.join(sorted(word))
    if key not in groups:
        groups[key] = []
    groups[key].append(word)
print("Skill groups:")
for group in groups.values():
    print(group)
```

```python
n = int(input("\nEnter number of packets: "))
packets = list(map(int, input("Enter packet IDs: ").split()))
numbers = set(packets)
longest = 0
for num in numbers:
    if num - 1 not in numbers:
        current = num
        length = 1
        while current + 1 in numbers:
            current += 1
            length += 1
        if length > longest:
            longest = length
print("Longest consecutive sequence:", longest)
```

```python
n = int(input("\nEnter number of appointments: "))
appointments = []
for i in range(n):
    start, end = map( int, input("Enter start and end time:").split()
    )
    appointments.append([start, end])
appointments.sort()
result = []
for start, end in appointments:
    if len(result) == 0:
        result.append([start, end])
    elif start <= result[-1][1]:
        result[-1][1] = max(result[-1][1], end)
    else:
        result.append([start, end])
print("Merged appointments:")
for appointment in result:
    print(appointment[0], appointment[1])
```
